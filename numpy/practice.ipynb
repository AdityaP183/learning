# Numpy Practice

Questions from: https://pynative.com/python-numpy-exercise/

```python
import numpy as np
```

Exercise 1: Create a 1D NumPy array of numbers from 0 to 9

```python
arr = np.arange(10)
print(arr)
```

Exercise 2: Convert 1D array to 2D

```python
reshaped_arr = np.reshape(arr, (2,5))

print("Original Array:", arr)
print("\nReshaped 2x5 Array:", reshaped_arr)
```

Exercise 3: Print Array Attributes

```python
my_array = np.array([[1, 2], [3, 4], [5, 6], [7, 8]], dtype=np.uint16)

print("The shape of the array is:", my_array.shape)
print("The number of dimensions is:", my_array.ndim)
print("The size of each element in bytes is:", my_array.itemsize)
```

Exercise 4: Create a 3×3 NumPy array of all True

```python
true_arrs = np.ones(shape=(3,3), dtype=bool)
print(true_arrs)
```

Exercise 5: Extract the documentation of NumPy’s arange() function

```python
print(np.info(np.arange))
```

Exercise 6: Create a 1D array filled with zeros and another filled with ones

```python
zeros_arr = np.zeros(5)
ones_arr = np.ones(5)

print("Zeros:", zeros_arr)
print("Ones:", ones_arr)
```

Exercise 7: Create a 1D array of 10 evenly spaced values between 5 and 50

```python
arr = np.arange(5, 51, 5)
print(arr)
```

Exercise 8: Convert a Python list into a NumPy array

```python
py_list = [1,2,3,4,5]

numpy_arr = np.array(py_list)

print("Python list: ", py_list)
print("Numpy list: ", numpy_arr)
```

Exercise 9: Find the memory size of a NumPy array of numbers from 0 to 9

```python
arr = np.arange(10)

print("Array:", arr)
print("Memory size in bytes:", arr.nbytes)
```

Exercise 10: Reverse a 1D NumPy array

```python
print(arr[::-1])
```

Exercise 11: Create a 3×3 identity matrix

```python
identity_matrix = np.eye(3)
print(identity_matrix)
```

Exercise 12: Create a 4×4 array and extract its first row and last column

```python
arr_4d = np.arange(1,17).reshape((4,4))

print("Array:", arr_4d)
print("First Row:", arr_4d[0])
print("Last Column:", arr_4d[:,-1])
```
