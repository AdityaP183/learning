1. Load the dataset into a DataFrame.

```python
df = pd.read_csv("../resources/world-happiness-report-2021.csv")
```

2. Display the first 10 rows of the dataset.

```python
print(df.head(10))
```

3. Find the number of rows and columns.

```python
dimensions = df.shape

print(f"Dataset Dimensions:\nRows: {dimensions[0]}\nColumns: {dimensions[1]}")
```

4. List all column names.

```python
print("Columns:\n")

for i, column in enumerate(df.columns):
    print(f"{i+1}: {column}")
```

5. Check the data type of each column.

```python
print(df.dtypes)
```

6. Identify columns that contain missing values.

```python
missing_columns = df.columns[df.isnull().sum() > 0]
print(missing_columns)
```

7. Count the total number of missing values in the dataset.

```python
print(df.isnull().sum().sum())
```

8. Display summary statistics for all numerical columns.

```python
df.describe()
```

9. Find the average happiness score across all countries.

```python
df["Ladder score"].mean()
```

10. Find the country with the highest happiness score.

```python
df.sort_values(by="Ladder score", ascending=False).head(1)
```

11. Find the country with the lowest happiness score.

```python
df.sort_values(by="Ladder score").head(1)
```

12. List the top 10 happiest countries.

```python
df.sort_values(by="Ladder score", ascending=False).head(10)
```

13. List the bottom 10 least happy countries.

```python
df.sort_values(by="Ladder score").head(10)
```

14. Find the average happiness score by region.

```python
df.groupby(["Regional indicator"])["Ladder score"].mean()
```

15. Find which region has the highest average happiness score.

```python
df.groupby(["Regional indicator"])["Ladder score"].mean().sort_values(ascending=False).head(1)
```

16. Find how many countries belong to each region.

```python
df.groupby(["Regional indicator"]).size()
```

17. Find the range (max − min) of happiness scores for each region.

```python
df.groupby(["Regional indicator"])["Ladder score"].agg(["min", "max"])
```

18. Find countries whose happiness score is above the global average.

```python
global_average_happiness_score = df["Ladder score"].mean()

df[df["Ladder score"] > global_average_happiness_score]["Country name"]
```

19. Sort countries by GDP per capita and list the top 10.

```python
df.sort_values("Logged GDP per capita", ascending=False).head(10)
```

20. Find the average life expectancy for countries with happiness score above 6.

```python
df[df["Ladder score"] > 6]["Healthy life expectancy"].mean().round()
```
